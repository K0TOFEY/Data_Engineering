```python
import pandas as pd
import sqlite3
```

```python
df = pd.read_excel("ШЧ_2_corr.xlsx")
conn = sqlite3.connect('transport.db')

df.to_sql("inspections", conn, if_exists='replace', index=False)
df.columns
```

```python
#тестовое задание

query = """
SELECT COUNT(*) AS total
FROM inspections
"""

print(pd.read_sql(query, conn))
```

```python
#задание 1

query = """
SELECT *
FROM inspections
WHERE "Причина" IS NOT NULL
"""

print(pd.read_sql(query, conn))
```

```python
#задание 2

query = """
SELECT * 
FROM inspections
WHERE DATE("Крайний срок") > '2019-04-15'
"""

print(pd.read_sql(query, conn))
```

```python
# задание 3

query = """
SELECT "Станция, перегон", COUNT(*) AS cnt
FROM inspections
WHERE "Статус" = 'ЗАКРЫТО'
GROUP BY "Станция, перегон"
ORDER BY cnt DESC
LIMIT 7
"""

print(pd.read_sql(query, conn))
```

```python
# задание 4

query = """
SELECT *, COUNT(*) AS cnt
FROM inspections
WHERE "Причина" IS NULL
GROUP BY "Классификация (Группа)"
ORDER BY cnt DESC
LIMIT 3
"""

print(pd.read_sql(query, conn))
```

```python
# задание 5

query = """
SELECT *
FROM inspections
WHERE DATE("Дата устранения") < DATE("Крайний срок")
"""

print(pd.read_sql(query, conn))
```

```python
#задание 6

query = """
SELECT * , COUNT(*) AS cnt
FROM inspections
WHERE "Станция, перегон" == "станция_0"
GROUP BY "Осмотр проводил"
ORDER BY cnt DESC
LIMIT 5
"""

print(pd.read_sql(query, conn))
```

```python
# задание 7

query = """
SELECT *
FROM inspections
WHERE DATE("Дата осмотра") BETWEEN '2019-05-01' AND '2019-06-01'
"""

print(pd.read_sql(query, conn))
```

```python
# задание 8

query = """
SELECT *
FROM inspections
WHERE 'Статус' == "УСТРАНЕН" AND 'Дата устранения' IS NULL
"""

print(pd.read_sql(query, conn))
```
